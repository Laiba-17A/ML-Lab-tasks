```python
from collections import Counter

class MajorityVoteClassifier:
    def __init__(self):
        self.majority_label = None

    def fit(self, X, y):
        counts = Counter(y)
        self.majority_label = counts.most_common(1)[0][0]
        return self

    def predict(self, X):
        return [self.majority_label for _ in X]


X_train = [
    ["Y", "N"],
    ["N", "Y"],
    ["Y", "Y"],
    ["Y", "N"],
    ["N", "Y"]
]

y_train = ["-", "-", "+", "-", "+"]

model = MajorityVoteClassifier()
model.fit(X_train, y_train)

X_test = [["Y", "Y"], ["N", "N"]]

print(model.predict(X_test))
```

```python
import random

class MemorizerClassifier:
    def __init__(self):
        self.memory = {}
        self.labels = []

    def fit(self, X, y):
        self.memory = {}
        self.labels = list(set(y))

        for features, label in zip(X, y):
            self.memory[tuple(features)] = label

        return self

    def predict(self, X):
        predictions = []

        for features in X:
            key = tuple(features)

            if key in self.memory:
                predictions.append(self.memory[key])
            else:
                predictions.append(random.choice(self.labels))

        return predictions


model = MemorizerClassifier()
model.fit(X_train, y_train)

print(model.predict(X_train))
```

```python
from collections import defaultdict, Counter

class DecisionStumpClassifier:
    def __init__(self, feature_index=0):
        self.feature_index = feature_index
        self.rules = {}
        self.default_label = None

    def fit(self, X, y):
        groups = defaultdict(list)

        for features, label in zip(X, y):
            value = features[self.feature_index]
            groups[value].append(label)

        self.rules = {}

        for value, labels in groups.items():
            self.rules[value] = Counter(labels).most_common(1)[0][0]

        self.default_label = Counter(y).most_common(1)[0][0]

        return self

    def predict(self, X):
        predictions = []

        for features in X:
            value = features[self.feature_index]

            predictions.append(
                self.rules.get(value, self.default_label)
            )

        return predictions


stump = DecisionStumpClassifier(feature_index=0)

stump.fit(X_train, y_train)

print(stump.rules)
print(stump.predict(X_test))
```

```python
class Node:
    def __init__(self, node_type, label=None, feature=None):
        self.type = node_type
        self.label = label
        self.feature = feature
        self.branches = {}


def predict(tree, x):
    current = tree

    while current.type != "leaf":
        feature = current.feature
        value = x[feature]
        current = current.branches[value]

    return current.label
```

```python
#build small tree manually
fail_leaf = Node("leaf", label="Fail")
pass_leaf = Node("leaf", label="Pass")

study_node = Node("internal", feature="Study")
study_node.branches["Low"] = fail_leaf
study_node.branches["High"] = pass_leaf

root = Node("internal", feature="Attendance")
root.branches["Low"] = fail_leaf
root.branches["High"] = study_node

student = {
    "Attendance": "High",
    "Study": "High"
}

print(predict(root, student))
```

```python
import math
#entropy
def entropy(labels):
    if len(labels) == 0:
        return 0

    counts = {}

    for label in labels:
        counts[label] = counts.get(label, 0) + 1

    total = len(labels)
    result = 0

    for count in counts.values():
        p = count / total
        result -= p * math.log2(p)

    return result
```

```python
dataset = [
    {"Attendance": "High", "Study": "High", "Assignment": "Yes", "Label": "Pass"},
    {"Attendance": "High", "Study": "Low", "Assignment": "Yes", "Label": "Pass"},
    {"Attendance": "High", "Study": "Low", "Assignment": "No", "Label": "Pass"},
    {"Attendance": "Low", "Study": "High", "Assignment": "Yes", "Label": "Pass"},
    {"Attendance": "Low", "Study": "Low", "Assignment": "No", "Label": "Fail"},
    {"Attendance": "Low", "Study": "Low", "Assignment": "Yes", "Label": "Fail"},
    {"Attendance": "High", "Study": "High", "Assignment": "No", "Label": "Pass"},
    {"Attendance": "Low", "Study": "High", "Assignment": "No", "Label": "Fail"}
]
```

```python
#mutual information
def conditional_entropy(dataset, feature, label_col="Label"):
    total = len(dataset)
    weighted_entropy = 0

    values = set(row[feature] for row in dataset)

    for value in values:
        subset = [
            row[label_col]
            for row in dataset
            if row[feature] == value
        ]

        weight = len(subset) / total
        weighted_entropy += weight * entropy(subset)

    return weighted_entropy


def mutual_information(dataset, feature, label_col="Label"):
    labels = [row[label_col] for row in dataset]

    h_y = entropy(labels)

    h_y_given_x = conditional_entropy(
        dataset, feature, label_col
    )

    return h_y - h_y_given_x
```

```python
#select the best features
features = ["Attendance", "Study", "Assignment"]

for feature in features:
    mi = mutual_information(dataset, feature)
    print(feature, mi)

best_feature = max(
    features,
    key=lambda f: mutual_information(dataset, f)
)

print("Best feature:", best_feature)
```

```python
#recursive training
from collections import Counter

def majority_label(dataset, label_col="Label"):
    labels = [row[label_col] for row in dataset]
    return Counter(labels).most_common(1)[0][0]


def train(dataset, features, label_col="Label"):

    labels = [row[label_col] for row in dataset]

    if len(set(labels)) == 1:
        return Node("leaf", label=labels[0])

    if len(features) == 0:
        return Node(
            "leaf",
            label=majority_label(dataset, label_col)
        )

    best_feature = max(
        features,
        key=lambda f: mutual_information(
            dataset, f, label_col
        )
    )

    node = Node("internal", feature=best_feature)

    values = set(row[best_feature] for row in dataset)

    remaining_features = [
        f for f in features if f != best_feature
    ]

    for value in values:
        subset = [
            row for row in dataset
            if row[best_feature] == value
        ]

        child = train(
            subset,
            remaining_features,
            label_col
        )

        node.branches[value] = child

    return node
```

```python
#train tree and make predictions
features = [
    "Attendance",
    "Study",
    "Assignment"
]

tree = train(dataset, features)

new_student = {
    "Attendance": "High",
    "Study": "Low",
    "Assignment": "No"
}

prediction = predict(tree, new_student)

print("Prediction:", prediction)
```

From-Scratch Implementation decision tree

```python
import math
from collections import Counter

def entropy(y):
    counts = Counter(y)
    n = len(y)
    return -sum((c/n) * math.log2(c/n) for c in counts.values())


def information_gain(y, left_y, right_y):
    parent = entropy(y)
    n = len(y)

    weighted = (
        (len(left_y)/n) * entropy(left_y)
        + (len(right_y)/n) * entropy(right_y)
    )

    return parent - weighted


def best_split(X, y):
    best = None
    best_gain = -1

    for feature in X.columns:
        for value in sorted(X[feature].unique()):
            left = y[X[feature] <= value]
            right = y[X[feature] > value]

            if len(left) == 0 or len(right) == 0:
                continue

            gain = information_gain(y, left, right)

            if gain > best_gain:
                best_gain = gain
                best = (feature, value)

    return best, best_gain

```

Simplified From-Scratch Random Forest

```python
from sklearn.tree import DecisionTreeClassifier
from collections import Counter
import numpy as np


class SimpleRandomForest:

    def __init__(self, n_trees=10, max_depth=4, random_state=42):
        self.n_trees = n_trees
        self.max_depth = max_depth
        self.random_state = random_state
        self.trees = []

    def fit(self, X, y):
        rng = np.random.default_rng(self.random_state)
        self.trees = []

        for _ in range(self.n_trees):

            indices = rng.choice(
                len(X),
                size=len(X),
                replace=True
            )

            tree = DecisionTreeClassifier(
                max_depth=self.max_depth,
                max_features="sqrt",
                random_state=int(
                    rng.integers(0, 1_000_000)
                )
            )

            tree.fit(X.iloc[indices], y.iloc[indices])
            self.trees.append(tree)

    def predict(self, X):
        predictions = np.array([
            tree.predict(X)
            for tree in self.trees
        ])

        return np.array([
            Counter(predictions[:, i]).most_common(1)[0][0]
            for i in range(X.shape[0])
        ])
```

Simplified Boosting Idea

```python
from sklearn.tree import DecisionTreeRegressor
import numpy as np

X_demo = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6]
])

y_demo = np.array([
    2,
    4,
    5,
    8,
    10,
    12
])

prediction = np.full(
    len(y_demo),
    y_demo.mean()
)

for stage in range(3):

    residual = y_demo - prediction

    learner = DecisionTreeRegressor(
        max_depth=1,
        random_state=stage
    )

    learner.fit(X_demo, residual)

    prediction = (
        prediction
        + 0.3 * learner.predict(X_demo)
    )

    print(
        "Stage:",
        stage + 1,
        "Predictions:",
        np.round(prediction, 2)
    )
```

Linear Regression — Important Theory
1. Regression
Regression is used when the target/label contains continuous values.
It is used to find the relationship between variables and predict future outcomes.
In regression, the output is a numerical/continuous value.
2. Linear Regression
Linear regression predicts the relationship between an independent variable (X) and a dependent variable (Y).
It assumes a linear relationship between the variables.
It finds the best-fit line by minimizing the error between actual and predicted values.
The goal is to find the line with the minimum error.
Formula
$$ Y = \beta_0 + \beta_1X $$

Where:

Y = dependent variable
X = independent variable
β₀ = intercept
β₁ = slope/coefficient
3. Simple Linear Regression
Has one independent variable and one dependent variable.
The model estimates:
Slope → change in dependent variable for one-unit change in independent variable.
Intercept → predicted value of Y when X = 0.
Formula
$$ Y = \beta_0 + \beta_1X $$
4. Best-Fit Line
The best-fit line is the line that has the least error.
Linear regression tries to find the best values of β₀ and β₁.
The difference between actual and predicted values is called the residual/error.
5. Residual

A residual is the difference between the observed value and predicted value.

$$ \epsilon_i = y_{predicted} - y_i $$

Where:

$$ y_{predicted} = \beta_0 + \beta_1X_i $$

6. Multiple Linear Regression
Used when there is one dependent variable and multiple independent variables.
Each independent variable has its own coefficient.
Formula
$$ Y = \beta_0 + \beta_1X_1 + \beta_2X_2 + ... + \beta_pX_p + \epsilon $$
Important Issues

Overfitting

Model becomes too complex when too many variables are added.
Can result in high training accuracy but low test accuracy.

Multicollinearity

Independent variables are highly/inter-related.

Feature Selection

Selecting the most useful predictors from many available features.

7. Linear Regression Coefficients

For:

$$ y = mx + b $$
y → dependent variable
x → independent variable
m → slope/coefficient
b → y-intercept

8. Regression Plot
A regression plot shows the relationship between two variables.
It helps visualize how well the linear regression line fits the data.
9. Assumptions of Linear Regression
1. Linearity
There should be a linear relationship between independent and dependent variables.
2. Independence
Observations should be independent of each other.
3. Homoscedasticity
The variance of residuals should remain constant across different levels of the independent variables.
4. Normality of Residuals
Residuals should be approximately normally distributed.
5. No Perfect Multicollinearity
Independent variables should not have a perfect linear relationship with each other.
6. No Autocorrelation
Residuals should not have a pattern or correlation across observations/time.
7. Additivity
The effect of each independent variable should be consistent, and total effects are the sum of individual effects.

10. Polynomial Regression
Polynomial regression models the relationship using a polynomial equation.
It can fit a curved line instead of only a straight line.
It can capture non-linear relationships.
Formula
$$ y = b_0 + b_1x + b_2x^2 + ... + b_nx^n + \epsilon $$
Important Points
Higher degree → more complex model.
Very high degree can cause overfitting.
Polynomial regression can capture non-linear patterns.
Degree can be selected using cross-validation.
Higher-degree models are harder to interpret.
Advantages
More flexible.
Can provide a better fit for certain data.
Disadvantages
Prone to overfitting.
More difficult to interpret.

13. Imbalanced Classification
Definition

An imbalanced dataset has substantially different numbers of examples in its classes.

Important Problem

Accuracy can be misleading.

For example:

990 legitimate
10 fraud

A model predicting everything as legitimate gets:

Accuracy = 99%
Recall for fraud = 0%

Therefore, high accuracy does not necessarily mean a good model for imbalanced data.

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, recall_score

dummy = DummyClassifier(strategy="most_frequent")

dummy.fit(X_train, y_train)

pred = dummy.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred))
print("Recall:", recall_score(y_test, pred, zero_division=0))
```

Practical Solutions

 Cross-Validation
Definition

Cross-validation repeatedly divides training data into training and validation portions to estimate performance on unseen data.

 K-Fold Cross-Validation

1. Divide data into K folds.
2. Train on K-1 folds.
3. Validate on the remaining fold.
4. Repeat until every fold has been used for validation.
5. Calculate mean/std of scores.

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.tree import DecisionTreeClassifier

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

model = DecisionTreeClassifier(random_state=42)

scores = cross_val_score(
    model,
    X,
    y,
    cv=kf,
    scoring="accuracy"
)

print("Fold Scores:", scores)
print("Mean:", round(scores.mean(), 3))
print("Std:", round(scores.std(), 3))
```

Data leakage occurs when information that should not be available during model training becomes available to the training process.

Grid Search
Tries all specified combinations of hyperparameters and selects the combination with the best cross-validation performance.

Random Search
Tries a limited number of randomly selected hyperparameter combinations instead of checking every combination.

TimeSeriesSplit
A cross-validation method for time-ordered data where training uses earlier observations and validation uses later observations.

Learning Curves
Graphs showing training and validation performance as training-data size changes, useful for detecting underfitting and overfitting.

Pipeline
Combines preprocessing and model training into one workflow, ensuring steps such as scaling are applied correctly during training and CV.

Data Leakage
Happens when information that should be unavailable during training accidentally enters the training process, giving unrealistically good results.

Search Cost
The computational time/resources required to test hyperparameter combinations. More combinations × more CV folds = higher search cost.

Choosing Scoring Metric
Selecting the metric used to decide which model or hyperparameter combination is better, such as accuracy/F1 for classification or R² for regression.
