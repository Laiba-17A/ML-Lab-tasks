```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

```

```python
X, y = make_blobs(n_samples=150, centers=3, random_state=42)

random_model = KMeans(n_clusters=3, init="random", n_init=10, random_state=42)
plus_model = KMeans(n_clusters=3, init="k-means++", n_init=10, random_state=42)

random_model.fit(X)
plus_model.fit(X)

print("Random Inertia:", random_model.inertia_)
print("K-Means++ Inertia:", plus_model.inertia_)
```

```python

plt.scatter(X[:, 0], X[:, 1], c=random_model.labels_, s=40)
plt.scatter(random_model.cluster_centers_[:, 0], random_model.cluster_centers_[:, 1], marker="X", s=200)
plt.title("Random Initialization")
plt.show()

plt.scatter(X[:, 0], X[:, 1], c=plus_model.labels_, s=40)
plt.scatter(plus_model.cluster_centers_[:, 0], plus_model.cluster_centers_[:, 1], marker="X", s=200)
plt.title("K-Means++ Initialization")
plt.show()
```
